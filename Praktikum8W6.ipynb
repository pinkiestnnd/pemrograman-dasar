```python
scores = [70, 65, 90, 85, 80]

total = 0                    # accumulator: starts at zero
count_pass = 0               # a counter is an accumulator that adds 1
for score in scores:
    total = total + score    # add each element
    if score >= 75:
        count_pass = count_pass + 1
    print("running total:", total)

average = total / len(scores)
print("Total   :", total)
print("Average :", average)
print("Scores >= 85:", count_pass, "of", len(scores))
```

# Soal No. 6G. Change the window : Edit the block: make it a 5-day moving average (start the loop at index 4, slice temps[i - 4 : i + 1], divide by 5). How many averages do you get for 7 values? Why?


```python
temps = [28, 31, 30, 33, 29, 27, 32]

# Daily change: start at index 1 because we look at i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"day {i + 1}: {temps[i]} °C, change {change:+d}")

# 3-day moving average: window = the 3 values ending at index i
avg3 = []
for i in range(2, len(temps)):
    window = temps[i - 2 : i + 1]
    avg3.append(round(sum(window) / 3, 1))

print("3-day averages:", avg3)
print(len(temps), "values ->", len(avg3), "averages")
```

# Soal NO. 6H. Explain the drop : Energy on day 17 fell by more than 40%. In groups, write down three operational reasons that could explain it and what data you would check to confirm each one. (Hint: look at run_hours[16] in the 30-day data section, and at the weekday block below.)



```python
energy = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# Daily change: start at index 1 because each value is compared with the one before
for i in range(1, len(energy)):
    change = (energy[i] - energy[i - 1]) / energy[i - 1] * 100
    if change < -20:
        print(f"Day {i + 1}: {change:+.1f}% versus the day before, investigate")

# 7-day moving average
avg_7 = []
for i in range(6, len(energy)):
    window = energy[i - 6 : i + 1]
    avg_7.append(sum(window) / 7)

print(f"{len(avg_7)} moving-average values ({len(energy)} days of data)")
print(f"Lowest {min(avg_7):,.0f} MWh/day, highest {max(avg_7):,.0f} MWh/day")
```

# Soal No. 6I. Your own array : Make a list temps with 7 daily temperatures (any values). Then: (1) show the first 3 days and the last 2 days with slicing; (2) add one day with append, then take it out again with pop; (3) show a sorted version with sorted without changing temps; (4) compute the average and find the highest temperature and which day it was; (5) build a new list of the temperatures above the average.



```python
import csv

# --- Langkah Tambahan: Membuat file CSV secara otomatis ---
raw_energy = [
    5577, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5535, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

with open("daily_production_synthetic.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["date", "energy_mwh", "run_hours"])
    for day, val in enumerate(raw_energy, start=1):
        # Hari ke-17 mengalami penurunan drastis sehingga jam operasinya berkurang
        hours = 14 if day == 17 else 24
        writer.writerow([f"2026-11-{day:02d}", val, hours])

# --- Program Asli Anda (Sudah Benar) ---
dates, energy, run_hours = [], [], []
with open("daily_production_synthetic.csv") as f:
    for row in csv.DictReader(f):
        dates.append(row["date"])
        energy.append(int(row["energy_mwh"]))
        run_hours.append(int(row["run_hours"]))

INSTALLED_MW = 330
DAY_NAMES = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"] # 1 November 2026 is a Sunday

# 1. Total, average, monthly capacity factor
total = sum(energy)
cf = total / (INSTALLED_MW * 24 * len(energy)) * 100
print(f"November energy : {total:,} MWh  |  average {total / len(energy):,.0f} MWh/day")
print(f"Capacity factor : {cf:.1f} %")

# 2. Weekdays vs weekends (day names repeat every 7 indices)
weekday, weekend = [], []
for i in range(len(energy)):
    if DAY_NAMES[i % 7] in ("Sat", "Sun"):
        weekend.append(energy[i])
    else:
        weekday.append(energy[i])
print(f"Weekday average {sum(weekday) / len(weekday):,.0f} vs weekend {sum(weekend) / len(weekend):,.0f} MWh")

# 3. Days without a full 24 hours of operation
for i in range(len(energy)):
    if run_hours[i] < 24:
        print(f"{dates[i]} ({DAY_NAMES[i % 7]}): only {run_hours[i]} h in operation, {energy[i]:,} MWh")

# 4. Text bar chart: one block = 100 MWh above 3,000 MWh (so the differences show)
print()
for i in range(len(energy)):
    print(f"{i + 1:>2} {DAY_NAMES[i % 7]} {'█' * ((energy[i] - 3000) // 100)} {energy[i]:,}")
```
