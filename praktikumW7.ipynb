# **Practicum 1: Basic functions**

```python
#Step 1.1: a report header procedure.

def print_header(title):
    """Procedure: print a report title between two lines."""
    line = "=" * (len(title) + 4)
    print(line)
    print("  " + title)
    print(line)


print_header("DAILY REPORT UNIT 1")
print("Energy on 2 November: 6098 MWh")
print()
print_header("WEEKLY SUMMARY")
```

```python
#Step 1.2: functions with parameters and return.

def mwh_to_gwh(energy_mwh):
    """Convert MWh to GWh (1 GWh = 1,000 MWh)."""
    return energy_mwh / 1000


def kcal_to_kj(heat_kcal):
    """Convert kcal to kJ (1 kcal = 4.184 kJ)."""
    return heat_kcal * 4.184


def average_power(energy_mwh, hours):
    """Average power (MW) = energy (MWh) divided by running time (hours)."""
    return energy_mwh / hours


print("Energy, day 2   :", mwh_to_gwh(6098), "GWh")
print("Calorific value :", round(kcal_to_kj(4400)), "kJ/kg")
print("Average power   :", round(average_power(6098, 24), 1), "MW")
print("Day 17 (18 h)   :", round(average_power(hours=18, energy_mwh=3592), 1), "MW")
```

```python
#Step 1.3: return is not print.

def heat_rate_print(coal_t, energy_mwh):
    print("heat rate:", round(coal_t * 4400 / energy_mwh))     # only prints


def heat_rate(coal_t, energy_mwh):
    return coal_t * 4400 / energy_mwh                          # returns the value


a = heat_rate_print(3258, 6098)
b = heat_rate(3258, 6098)
print("a =", a)
print("b =", round(b))
print("Efficiency from b:", round(860 / b * 100, 2), "%")
```

```python
#Step 1.4: returning several values.

def summary(data):
    """Return (lowest, highest, average) of an array."""
    return min(data), max(data), sum(data) / len(data)


energy = [5576, 6098, 6078, 6095, 6074, 6077, 5564]     # MWh, days 1-7

lowest, highest, average = summary(energy)
print(f"Lowest   : {lowest} MWh (day {energy.index(lowest) + 1})")
print(f"Highest  : {highest} MWh (day {energy.index(highest) + 1})")
print("Average  :", round(average, 1), "MWh")
print("One value:", summary([7920]))
```

# **Practicum 3: Modularizing the plant calculation**

```python
#Step 3.1: four formulas, four functions, four tests.

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh) from coal (t) and energy sent out (MWh)."""
    if energy_mwh <= 0:
        raise ValueError("energy_mwh must be positive")
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate. 1 kWh = 860 kcal."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) for a period."""
    return energy_mwh / (installed_mw * hours) * 100


def specific_coal_consumption(coal_t, energy_mwh):
    """Specific coal consumption (t/MWh)."""
    return coal_t / energy_mwh


# Tests against hand calculations
assert heat_rate(1, 2) == 2200                          # 1 x 4400 / 2
assert round(thermal_efficiency(2150), 2) == 40         # 860 / 2150 = 0.4
assert round(capacity_factor(7920)) == 100              # 330 MW x 24 h = 7920 MWh
assert specific_coal_consumption(100, 200) == 0.5       # 100 t / 200 MWh
print("All four tests passed")

hr = heat_rate(3258, 6098)
print(f"Day 2: HR {hr:,.0f} kcal/kWh, efficiency {thermal_efficiency(hr):.2f} %, "
      f"CF {capacity_factor(6098):.1f} %, SCC {specific_coal_consumption(3258, 6098):.3f} t/MWh")
```

```python
#Step 3.2: return inside branches

def efficiency_status(efficiency):
    """'POOR' if < 36.0 %, 'LOW' if < 36.3 %, otherwise 'NORMAL'."""
    if efficiency < 36.0:
        return "POOR"
    if efficiency < 36.3:
        return "LOW"
    return "NORMAL"


for e, expected in [(35.9, "POOR"), (36.0, "LOW"), (36.29, "LOW"), (36.3, "NORMAL"), (36.5, "NORMAL")]:
    assert efficiency_status(e) == expected, f"{e} should be {expected}"
    print(f"{e:>5} % -> {efficiency_status(e)}")
print("All limits correct")
```

```python
#Step 3.3: the main program only calls functions.

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) for a period."""
    return energy_mwh / (installed_mw * hours) * 100


def efficiency_status(efficiency):
    """'POOR' if < 36.0 %, 'LOW' if < 36.3 %, otherwise 'NORMAL'."""
    if efficiency < 36.0:
        return "POOR"
    if efficiency < 36.3:
        return "LOW"
    return "NORMAL"


def print_row(day, hr, eff, cf, status):
    """Procedure: one row of the report table."""
    print(f"{day:>3}  {hr:>7,.0f}  {eff:>6.2f}  {cf:>5.1f}  {status}")


# --- Data Unit 1, 14-20 November 2026 ---
first_day = 14
energy = [5533, 5566, 6097, 3592, 6059, 6100, 6098]      # MWh
coal = [2996, 3009, 3257, 1950, 3234, 3261, 3262]        # t

# --- Main program ---
print(f"{'Day':>3}  {'HR kcal':>7}  {'Eff %':>6}  {'CF %':>5}  Status")
efficiencies = []
for i in range(len(energy)):
    hr = heat_rate(coal[i], energy[i])
    eff = thermal_efficiency(hr)
    efficiencies.append(eff)
    print_row(first_day + i, hr, eff, capacity_factor(energy[i]), efficiency_status(eff))

lowest = min(efficiencies)
print(f"\nAverage efficiency : {sum(efficiencies) / len(efficiencies):.2f} %")
print(f"Lowest efficiency  : {lowest:.2f} % (day {first_day + efficiencies.index(lowest)})")
print(f"Weekly CF          : {capacity_factor(sum(energy), hours=24 * len(energy)):.1f} %")
```

```python
#Step 3.4: reuse for another unit.

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) for a period."""
    return energy_mwh / (installed_mw * hours) * 100


def unit_report(name, installed_mw, gcv, energy, coal):
    """Procedure: KPI summary of one unit from energy (MWh) and coal (t) arrays."""
    total_energy = sum(energy)
    hr = heat_rate(sum(coal), total_energy, gcv_kcal_kg=gcv)
    cf = capacity_factor(total_energy, installed_mw=installed_mw, hours=24 * len(energy))
    print(f"{name}: {len(energy)} days, HR {hr:,.0f} kcal/kWh, "
          f"efficiency {thermal_efficiency(hr):.2f} %, CF {cf:.1f} %")


unit_report("Unit 1", 330, 4400,
            [5576, 6098, 6078, 6095, 6074, 6077, 5564],
            [2998, 3258, 3250, 3263, 3249, 3246, 2993])
unit_report("Unit 3", 100, 4200,
            [2210, 2235, 2190],
            [1270, 1281, 1262])
```
