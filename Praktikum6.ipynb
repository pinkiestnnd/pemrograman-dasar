```python
#No. 1 kelipatan 5, jika angka habis dibagi 5
angka = int(input("Masukkan angka: "))

if angka % 5 == 0:
    print("Kelipatan 5")
```

```python
#No. 2 menentukan tahun kabisat dengan menggunakan kondisi:
tahun = int(input("Masukkan tahun: "))
 #syarat tahun kabisat:
if tahun % 400 == 0 or (tahun % 4 == 0 and tahun % 100 != 0):
    print(tahun, "adalah tahun kabisat") #jika habis dibagi 400-> Kabisat
else:
    print(tahun, "bukan tahun kabisat") #habis dibagi 4 dan tidak habis dibagi 100 -> kabisat
```
