```python
# Soal No. 5 Katup Pengaman (20 poin | CLO-1 | TP-3, TP-4)
status_manual_override = False

def cek_katup(tekanan_uap, status_manual_override):
    if tekanan_uap > 150 and status_manual_override == False:
        return "Katup Terbuka"
    elif status_manual_override == True:
        return "Katup Mode Manual"
    else:
        return "Katup Tertutup"

# Uji kelima kasus
kasus = [(160, False), (160, True), (140, False), (150, False), (140, True)]
for tekanan, override in kasus:
    print(f"{tekanan} | {override} -> {cek_katup(tekanan, override)}")
```

1. Fungsi cek_katup: tempat aturan katup ditulis. Fungsi menerima tekanan dan status override, lalu memberi hasil berupa status katup.

2. Aturan if-elif-Else:
Katup terbuka jika tekanan lebih dari 150 dan override False.
Katup mode manual jika override True, berapa pun tekanannya.
Selain itu, katup tertutup.

3. Daftar kasus berisi 5 kombinasi dari tabel soal

4. Perulangan for: mengambil kombinasi satu per satu, mengujinya dengan fungsi, lalu mencetak hasilnya.

5. hasilny menunjukan hasil dari eliminasi 5 penguji kasus seperti yang ditabel.

```python
#Challenge: Pemantauan Beban Listrik Harian

# MASALAH: memantau beban listrik harian (kWh) dan mendeteksi hari yang boros.
# CARA KERJA: data sensor berupa teks. Setiap data diubah ke angka,
# data rusak dilewati, lalu tiap hari diberi status Hemat/Normal/BOROS.

hari = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"]
beban_harian = ["12.5", "14.0", "ERR", "20.3", "9.8", "16.7", "25.1"]

data_valid = []

for nama, teks in zip(hari, beban_harian):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"{nama}: data tidak valid, dilewati")
        continue

    data_valid.append(nilai)

    if nilai > 20:
        status = "BOROS"
    elif nilai >= 10:
        status = "Normal"
    else:
        status = "Hemat"

    print(f"{nama}: {nilai} kWh -> {status}")

total = sum(data_valid)
rata_rata = total / len(data_valid)
print(f"Total: {total:.1f} kWh")
print(f"Rata-rata: {rata_rata:.2f} kWh")
```

Masalah: memantau konsumsi listrik harian dan menandai hari yang boros.
list: data_valid menyimpan hanya data yang berhasil dibaca.
for + zip: mengulang nama hari dan datanya secara berpasangan.
try-except: menangani sensor rusak ("ERR") agar program tidak error, cukup dilewati.
if-elif-else: menentukan status: di atas 20 kWh BOROS, 10 sampai 20 Normal, di bawah 10 Hemat.
Akhir: sum() menghitung total, lalu dibagi len() untuk rata-rata. Rata-rata hanya dari 6 data valid, bukan 7.
