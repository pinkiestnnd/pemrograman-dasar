```python
#Nadia Putri Affanur Jabar (4232501023)
# Soal No. 5 Katup Pengaman (20 poin | CLO-1 | TP-3, TP-4)
status_manual_override = False

def cek_katup(tekanan_uap, status_manual_override):
    if tekanan_uap > 150 and status_manual_override == False:
        return "Katup Terbuka"
    elif status_manual_override == True:
        return "Katup Mode Manual"
    else:
        return "Katup Tertutup"

# Uji kelima kasus
kasus = [(160, False), (160, True), (140, False), (150, False), (140, True)]
for tekanan, override in kasus:
    print(f"{tekanan} | {override} -> {cek_katup(tekanan, override)}")
```

Program ini menentukan status katup pengaman berdasarkan tekanan uap dan status manual override.

Fungsi cek_katup dibuat untuk menampung aturan penentuan status katup. Fungsi menerima dua nilai, yaitu tekanan uap dan status manual override, lalu mengembalikan teks status katup.
Kondisi pertama (if) mengecek dua syarat sekaligus dengan operator and: tekanan harus lebih dari 150 psi dan status manual override harus False. Jika keduanya benar, hasilnya “Katup Terbuka”. Karena memakai tanda lebih dari (bukan lebih dari atau sama dengan), tekanan tepat 150 tidak membuat katup terbuka.
Kondisi kedua (elif) mengecek apakah status manual override bernilai True. Jika ya, hasilnya “Katup Mode Manual”, berapa pun nilai tekanannya.
Kondisi ketiga (else) berlaku untuk semua kasus selain dua di atas, hasilnya “Katup Tertutup”.
List kasus berisi lima kombinasi tekanan dan status override yang sama dengan tabel pada soal, sehingga semuanya bisa diuji sekaligus.
Perulangan for mengambil kombinasi satu per satu, mengirimnya ke fungsi cek_katup, lalu mencetak hasilnya dengan f-string.

```python
#Nadia Putri affanur Jabar (4232501023)
#Challenge: Pemantauan Beban Listrik Harian

# MASALAH: memantau beban listrik harian (kWh) dan mendeteksi hari yang boros.
# CARA KERJA: data sensor berupa teks. Setiap data diubah ke angka,
# data rusak dilewati, lalu tiap hari diberi status Hemat/Normal/BOROS.

hari = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"]
beban_harian = ["12.5", "14.0", "ERR", "20.3", "9.8", "16.7", "25.1"]

data_valid = []

for nama, teks in zip(hari, beban_harian):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"{nama}: data tidak valid, dilewati")
        continue

    data_valid.append(nilai)

    if nilai > 20:
        status = "BOROS"
    elif nilai >= 10:
        status = "Normal"
    else:
        status = "Hemat"

    print(f"{nama}: {nilai} kWh -> {status}")

total = sum(data_valid)
rata_rata = total / len(data_valid)
print(f"Total: {total:.1f} kWh")
print(f"Rata-rata: {rata_rata:.2f} kWh")
```

Program ini memantau konsumsi listrik harian dalam kWh dan menandai hari yang boros.

Dua list dibuat: satu berisi nama hari, satu berisi data beban dalam bentuk teks (ada satu data rusak). List kosong data_valid disiapkan untuk menyimpan data yang berhasil dibaca.
Fungsi zip menggabungkan dua list menjadi pasangan, sehingga setiap putaran mendapat satu hari beserta datanya.
Blok try-except mengubah teks menjadi angka. Jika gagal, program mencetak pesan data tidak valid dan melewati hari tersebut agar tidak terjadi error.
Data yang valid dimasukkan ke list data_valid dengan method append.
Percabangan if-elif-else menentukan status: lebih dari 20 kWh berarti BOROS, 10 sampai 20 kWh berarti Normal, dan di bawah 10 kWh berarti Hemat.
Program mencetak hari, besar beban, dan statusnya.
Setelah perulangan, fungsi sum menjumlahkan semua data valid, lalu hasilnya dibagi jumlah data (dihitung dengan len) untuk mendapat rata-rata. Rata-rata hanya dihitung dari data valid, bukan semua data.
Total dan rata-rata dicetak dengan format angka desimal yang rapi.
