```python
#Praktikum 5 week 7
#Nadia Putri affanur Jabar
#4232501023

#Soal L1 — Pola Angka Segitiga
#Buat program yang mencetak pola angka berikut menggunakan nested loop. Setiap baris mencetak angka dari 1 hingga nomor baris tersebut.


#print("--- Soal L1: Pola Angka ---")
n = int(input("Masukkan jumlah baris: "))
for baris in range(1, n + 1):
    for angka in range(1, baris + 1):
        print(angka, end=" ")
    print()
```

```python
#Soal L2 — Tabel Data Sensor Pembangkit
#Seorang teknisi memiliki data pembacaan daya (MW) dari 3 unit pembangkit selama 4 hari:
data_daya = [
    [142, 138, 145, 140],  # Unit A
    [115, 120, 118, 122],  # Unit B
    [ 95,  98,  92,  96],  # Unit C
]
nama_unit = ["Unit A", "Unit B", "Unit C"]

print("--- Soal L2: Rata-rata Daya per Unit ---")
for i in range(len(data_daya)):         # iterasi setiap unit
    total = 0
    for daya in data_daya[i]:           # iterasi setiap hari pada unit ke-i
        total += daya
    rata2 = total / len(data_daya[i])   # bagi dengan jumlah hari
    print(f"{nama_unit[i]}: {rata2:.1f} MW")
```

```python
#Soal L3 — Gabungan while + Nested Loop: Simulasi Pencarian Anomali
#Sebuah sistem monitoring terus membaca data suhu dari 4 sensor setiap siklus. Program harus:
print("--- Soal L3: Monitoring Sensor ---")
import random
random.seed(42)

siklus = 0
ada_anomali = True

while ada_anomali:                       # jalankan selama masih ada anomali
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    for sensor_id in range(1, 5):        # 4 sensor
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:       # cek kondisi anomali
            print(f"  ⚠ Sensor {sensor_id}: {suhu}°C — ANOMALI")
            ada_anomali = True
        else:
            print(f"  ✓ Sensor {sensor_id}: {suhu}°C — Normal")

print(f"\nSemua sensor normal pada siklus ke-{siklus}.")
```
