```python
#Konversi Total Detik
total_detik = int(input("Masukkan total detik : "))
jam         = total_detik // 3600
sisa_detik  = total_detik % 3600
menit       = sisa_detik // 60
detik       = sisa_detik % 60
print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
#Nilai Akhir dan Kelulusan
nilai_mahasiswa   = float(input("Masukkan nilai mahasiswa : "))
standar_kelulusan = float(input("Masukkan nilai standar   : "))
apakah_lulus  = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan
print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")

#Modifikasi 2: nilai dari 5 variabel (bobot buatan sendiri)
tugas     = float(input("Nilai tugas     : "))
kuis      = float(input("Nilai kuis      : "))
praktikum = float(input("Nilai praktikum : "))
uts       = float(input("Nilai UTS       : "))
uas       = float(input("Nilai UAS       : "))
nilai_mahasiswa = round((0.2*tugas) + (0.1*kuis) + (0.2*praktikum) + (0.2*uts) + (0.3*uas), 2)
standar_kelulusan = float(input("Standar kelulusan : "))
print(f"Nilai akhir: {nilai_mahasiswa}")
print(f"Lulus? {nilai_mahasiswa >= standar_kelulusan}")
print(f"Perlu remedial? {nilai_mahasiswa < standar_kelulusan}")
```

```python
#Praktik 5: Diskon Belanja
belanja     = int(input("Masukkan total belanja : "))
diskon      = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon
print("Total harga setelah diskon =", total_harga)
```

```python
#Latihan 1: Tahun Kabisat
tahun = int(input("Masukkan tahun : "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
print("Apakah", tahun, "tahun kabisat?", kabisat)
```

```python
#Latihan 2: Program Buatan Sendiri
nama   = str(input("Nama pelanggan        : "))
harga  = float(input("Harga per porsi (Rp)  : "))
jumlah = int(input("Jumlah porsi          : "))

subtotal = harga * jumlah
ppn      = subtotal * 0.11
total    = subtotal + ppn

print("\n=== STRUK PEMBELIAN ===")
print("Pelanggan : %s" % nama)
print("Subtotal  : Rp%.0f" % subtotal)
print("PPN 11%%   : Rp%.0f" % ppn)
print("Total     : Rp%.0f" % total)

bayar = float(input("Uang bayar (Rp)      : "))
print(f"Kembalian : Rp{bayar - total:.0f}")
```
